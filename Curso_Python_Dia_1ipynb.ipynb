### Operaciones matematicas

```python
# 4 nivel de importancia
# suma ordinaria
print(3+4)
variable1,variable2 = 20,40
print(variable1 + variable2)

# resta ordinaria
print(2 - 5)
v1 = 10.4
v2 = 15
resultado = v1 - v2
print(resultado)
```

```python
# 3er nivel de importancia
# Multiplicacion ordinaria
print("Multiplicación ordinaria")
M1,M2 = 2, 3.14
print(M1 * M2)
# Division ordinaria
print("-----------------------")
print("División ordinaria")
M1,M2,M3,M4,M5,M6 = 2,5,6,5,4,2
print(M1 / M2)
print(M3 / M4)
print(M5 / M6)
# Division de piso
print("-----------------------")
print("División de piso")
print(M1 // M2)
print(M3 // M4)
print(M5 // M6)
```

```python
# Potencia
print("Potencia")
print(2 ** 3)
P1,P2 = 7,2
print(P1 ** P2)
```

```python
# Area de un circulo
pi = 3.14
r = 4
A = pi * r ** 2
print(pi)
A
```
