## Que es un modulo?
Un modulo es una unidad de organización del código que permite encapsular y dividir el software en partes más manejables, generalmente representado como un archivo que contiene codigo relacionado, como funciones, variables (constantes) o metodos de clase.

JSON es un formato para guardar y transmitr datos. Es muy usado en APIs, archivos de configuración y almacenamiento de información.
(Se parece a un diccionario de python pero es solo texto)

para que sirve?
  -Sirve para convertir objetos de python
  -> texto JSON y también texte JSON -> objetos de python

  Esto se llama:
    -Serializar: convertir un objeto de python en texto JSON
    -Deserializar: convertir un texto de JSON a un objeto usable en python

```python
# Funciones principales:
# 1) json.dums(obj): convierte un diccionario u objeto de python en un string JSON
import json
obj = {"nombre" : "Ana"
, "edad" : 25}
print(type(obj))
s = json.dumps(obj, ensure_ascii= False)
print(s)
print(type(s))
```

```python
# json.loads(texto_json): convierte un strin JSON en un diccionario python
restaurar = json.loads(s)
print(restaurar)
print(type(restaurar))
# Al venir de una pagina/persona/archivo desconocido debes de tener cuidado

va = '''
Pueden contener:
datos malinformados
información inesperada
campos faltantes
datos demasiado grandes
variables incorrectas que rompan tu programa
'''
print(va)
```

## Numpy
provee el tipo de dato ndarray, operaciones vectorizadas, brodcasting, utilidades de algebra lineal y funciones matematicas en arrays. es la base sobre la que se apoyan muchas librerias ciencitificas en python

### nota IMPORTANTE: el arreglo de numpy no es igual a la lista de python, este no puede tener diferentes tipos de datos en el mismo

```python
#pip install numpy
import numpy as np
A = np.array([[1,2,3],[4,5,6]]) #array (3,3)
v = np.array([10,20,30]) #array (3)
print(A + v)
print("----------------")
prueba = [1,2,3]
prueba2 = [1,2,3]
print (prueba + prueba2)
```

```python
# mas funciones de numpy
z = np.zeros((3,3)) #array (3,3)
print(z)
o = np.ones((2,4))
r = np.random.rand(3,3)
```

```python
# rangos numericos
nums  = np.arange(0,10,2) #saca los valores entre 0 y 10 de 2 en 2
print(nums)
x = np.linspace(0,1,5) #saca 5 valores espacciados de manera "equitativa" entre 0 y 1
print(x)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** b)
```

```python
print(a.dot(b)) #producto entre  dos matrices
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[10,20],[30,40],[50,60]])
#1 * 10 + 2*30  + 3 * 50
#
print(c.dot(d))
```

```python
data = np.array([1,2,3,4,5])
print(data.min()) #minimo
print(data.max()) #maximo
print(data.sum()) #suma
print(data.mean()) #promedio
print(data.std()) #desviacion estandar
```

```python
A = np.array([1,6,3,9,2])
print(A[A > 5])
```
