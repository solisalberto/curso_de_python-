```python
def saludar(nombre):
  return f"Hola, {nombre}!"

def despedirse(nombre):
  return f"Adiós, {nombre}!"
def mensaje_completo(nombre):
  saludo = saludar(nombre)
  despedida = despedirse(nombre)
  return saludo + "\n" + despedida
print(mensaje_completo("Victor"))
```

```python
# Ahora supongamos que tenemos una heladeria
def pedir_sabor():
  sabores = ["vainilla", "chocolate", "fresa"]
  print("Sabor disponible", "," ,sabores)
  # lower () hace minusculas las palabras
  # strip () borra los espacios
  sabor = input("Elige un sabor: ").strip().lower()

  while sabor not in sabores:
    sabor = input("Elige un sabor: (Ingresa un sabor valido) ").strip().lower()
  return sabor

def pedir_cantidad():
  cantidad = input("Cuantas bolas de nieves quieres?: ")
  while cantidad <=0:
    cantidad = input("Cuantas bolas de nieves quieres?: (Ingresa un valor valido)")
  return cantidad

def pedir_helado():
  sabor = pedir_sabor()
  cantidad = pedir_cantidad()
  return f"Pediste {cantidad} bolas de {sabor}"
  return cantidad, sabor

total_pedidos = int(input("Cuantos clientes antenderas"))
for i in range(total_pedidos):
  print(pedir_helado())
```

## funciones lambda
Una funcion lambda es una función acortada para escribir una funcion

de esta forma
nombreFuncion = lambda parametro: acción

la diferencia entre las funciones lambda y las funciones normales

```python
sumar = lambda x,y : x + y
print(sumar(1,2))
```

```python
descuentoLambda = lambda desc, precio: ((precio*(1-desc)), (precio * desc))
nuevoPrecio = descuentoLambda(0.15, 1000)
print("El precio a pagar es: ",nuevoPrecio[0], "el descuento que se hizo fue de: ", nuevoPrecio)
print(type(nuevoPrecio))
```

```python
def promediosYmayor(promedios):
  promedio = 0
  for x in promedios:
    promedio += x
  promedio = promedio / len(promedios)
  return promedio, max(promedios)

promedios = []
continuar = ""

while True:
  if len(promedios) > 0:
    continuar = ""
    while continuar not in ("si", "no"):
      continuar = input("Ingrese una respuesta válida: si/no").lower()
    if continuar == "no":
      break

  while True:
    try:
      calif = int(input("Indique la calificación: "))
      if calif < 0 or calif > 100:
        print("Ingrese un valor valido")
      else:
        promedios.append(calif)
        break
    except ValueError:
      print("Ingrese un valor numérico")

prom, mayor = promediosYmayor(promedios)
print("\nResultados:")
print("Promedio:", prom)
print("Mayor:", mayor)
```

```python
# calculadora -> suma, resta, producto, division y potencia
def suma(a,b):
  return a + b
def resta(a,b):
  return a - b
def producto(a,b):
  return a * b
def division(a,b):
  if b == 0:
    return "Error: no se puede dividr entre 0"
  return a / b
def potencia(a,b):
  return a ** b

def menu():
  print("-----CALCULADORA------")
  print("1. Suma")
  print("2. Resta")
  print("3. Producto")
  print("4. División")
  print("5. Potencia")
  print ("6. Salir")

while True:
  menu()
  try:
    op = int(input("Seleccione una opción:"))
    if op < 1 or op > 6:
      print("Ingrese una opción válida")
    if op == 6:
      print("SALIENDO....")
  except ValueError:
    print("Error: No ingreso una opción válida")
  break

  try:
      a = int(input("Ingrese el primer número: "))
      b = int(input("Ingrese el segundo número: "))
  except ValueError:
      print("Ingrese un valor numérico")
      if op == 1:
        print("La suma es: ", suma(a,b))
      elif op == 2:
        print("La resta es: ", resta(a,b))
      elif op == 3:
        print("El producto es: ", producto(a,b))
      elif op == 4:
        print("La división es: ", division(a,b))
      elif op == 5:
        print("La potencia es: ", potencia(a,b))
      else:
        print("Opción inválida")
```
