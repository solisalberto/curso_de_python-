```python
cal = 80
if cal == 100:
  print("A+")
elif cal>=90:
  print("A")
elif cal>=80:
  print("B")
elif cal>=70:
  print("C")
elif cal>=60:
  print("D")
else:
  print("F")
```

```python
cal = 80
if cal >=0 and cal <=100:
  if cal == 100:
    print("A+")
  elif cal>=90:
    print("A")
  elif cal>=80:
    print("B")
  elif cal>=70:
    print("C")
  elif cal>=60:
    print("D")
  else:
    print("F")
else:
  print("Error: La calificación debe de estar entre 0 y 100")
```

#While

```python
suma = 0
num = int(input("Dame un numero: "))


while num != 0:
  suma += num
  num = input("Dame un numero: ")
  num = int(num)

print(f"La suma es {suma}")
```

```python
print("Juego: Adivina el Número")
numero_secreto = int(input("Jugador 1, escribe el número secreto (1-20): "))
intentos = 0
max_intentos = 4
while intentos < max_intentos:
  intento = int(input("Adivina el número: "))
  intentos += 1
  if intento == numero_secreto:print("¡Correcto!")

  break
else:print("Incorrecto, intenta de nuevo.")

```
