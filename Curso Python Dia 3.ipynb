# Estructuras de datos
Las estructuras de datos son formas de organizar información para trabajar con ellas de manera eficiente.
- Listas -> ordenadas, modificadas, permiten duplicadas
- Tuplas -> ordenadas, inmutables, permiten duplicados
- sets -> no son ordenados, no se permiten duplicados
- diccionarios -> pares por clave:valor

Nota:
- Pilas first int, last out
- Colas first int, first out
- Lístas enlazadas

### Juego: Piedra, Papel o Tijeras (Corregido)

He corregido la indentación del código y he añadido `.lower()` a la entrada del usuario para que el juego sea insensible a mayúsculas y minúsculas. Así, 'Piedra' o 'piedra' se tratarán de la misma manera.

```python
import random

opciones = ["piedra", "papel", "tijeras"]

while True:  # Bucle para jugar múltiples rondas
    while True:  # Bucle para obtener una entrada válida del usuario
        eleccion = input("\nElige piedra, papel o tijeras (o escribe salir si ya no quieres jugar): ").lower()
        if eleccion not in opciones and eleccion != "salir":
            print("Opción no válida, intenta de nuevo.")
        else:
            break

    if eleccion == "salir":
        print("¡Gracias por jugar!")
        break

    aleatoria = random.choice(opciones)
    print("La computadora eligió:", aleatoria)

    # Determinamos el ganador
    if eleccion == aleatoria:
        print("Empate")
    elif (eleccion == "piedra" and aleatoria == "tijeras") \
            or (eleccion == "papel" and aleatoria == "piedra") \
            or (eleccion == "tijeras" and aleatoria == "papel"):
        print("¡Ganaste!:))")
    else:
        print("Perdiste:((")
```

```python
matutino = {20322,22183,123123,53234}
vespertino = {20322,11111,22222,33333}
total_matriculas = matutino.union(vespertino)
print(total_matriculas)
total_matriculas = matutino | vespertino
print(total_matriculas)
```

```python
# Intersección
ca = {1234,121,646,12414,73434,124124,675234}
al = {1234, 111, 222, 333, 444}
ca_al = ca.intersection(al)
print(ca_al)
ca_al = ca & al
print(ca_al)
```

```python
# Diferencia
solo_cal = ca.difference(al)
print(solo_cal)
```

##Diccionarios
Los diccionarios son colecciones no ordenadas de pares clave:valor

```python
alumno = {
    #clave  valor
    "nombre": "Elliud Sanchez",
    "edad" : 21,
    "carrera": "lcc" ,
    "semestre": 7,
    "promedio": 88,
    "calificaciones": [100,94]
}
print(type(alumno))
alumno
```

```python
print("El alumno", alumno["nombre"], "esta estudiando", alumno["carrera"])
```

```python
alumno["area preferida"] = "Ciencia de Datos"
alumno
```

```python
for valor in alumno:
  print(alumno[valor])
```

```python
for clave, valor in alumno.items():
  print(clave,":", valor)
```

```python
# tambien el metodo .clear() se puede usar aqui
# tenemos el get que nos ayuda a obtener el valor value() de cierta clva (key)
print(alumno.get("nombre"))
print(alumno.values)
```

```python
alumno_1 = alumno.items()
print(alumno_1)
```

```python
# el metodo .values() sirve para obtener los valores del diccionario, tambien se puede usar en iteración
alumno_v = alumno.values()
print(type(alumno_v))
print(alumno_v)
```

```python
if "carrera" in alumno:
  print("Si existe")
else:
  print("No existe")

if "dinero" in alumno:
  print("Alumno tiene dinero")
else:
  print("Alumno no tiene dinero")
```

```python
# .copy puedes copiar todo el diccionario en una nueva variable
# update -> sirve para cambiar los valores
alumno.update({"semestre":6})
alumno
```
